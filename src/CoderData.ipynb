```python
p_set = set(beataml.proteomics.improve_sample_id.unique().tolist())
t_set = set(beataml.transcriptomics.improve_sample_id.unique().tolist())
d_set = set(merged_df.improve_sample_id.unique().tolist())
complete_samples = list(p_set & t_set & d_set)
print(f"data counts: {len(d_set)}, complete data counts: {len(complete_samples)}")
```

```python
def extract_sample_info(ref_df: pd.DataFrame, index_ref: dict, info_name: str, sample_id: str, mean_values: list, std_values: list, mean_fill:int=0, std_fill:int=0.0001) -> None:
    cands = ref_df[(ref_df["improve_sample_id"]==sample_id)]
    for entrez in cands["entrez_id"].unique().tolist():
        values = ref_df[(ref_df["entrez_id"]==entrez)][info_name].values
        if values.size==0:
            mean_values[index_ref[entrez]] = mean_fill
            std_values[index_ref[entrez]] = std_fill
        else:
            mean_values[index_ref[entrez]] = np.mean(values)
            std_values[index_ref[entrez]] = max(np.std(values), std_fill)
    
```

```python
import time
t_mean_data = []
t_std_data = []
p_mean_data = []
p_std_data = []
unique_entrez_ids = list(set(beataml.proteomics.entrez_id.unique()).union(set(beataml.transcriptomics.entrez_id.unique())))
entrez_length = len(unique_entrez_ids)
print(f"info dimension {entrez_length}")
entrez_lookup = {element: index for index, element in enumerate(unique_entrez_ids)}

count = 0
print("Start")
for improve_sample_id in complete_samples:
    transcriptomics_mean = [None]*entrez_length
    transcriptomics_std = [None]*entrez_length
    proteomics_mean = [None]*entrez_length
    proteomics_std = [None]*entrez_length
    start = time.time()
    extract_sample_info(beataml.transcriptomics, entrez_lookup, "transcriptomics", improve_sample_id, transcriptomics_mean, transcriptomics_std)
    end = time.time()
    print(f"Transcriptomics Execution time: {end - start:.4f} seconds")
    start = time.time()
    extract_sample_info(beataml.proteomics, entrez_lookup, "proteomics", improve_sample_id, proteomics_mean, proteomics_std)
    end = time.time()
    print(f"Proteomics Execution time: {end - start:.4f} seconds")
    t_mean_data.append(transcriptomics_mean)
    t_std_data.append(transcriptomics_std)
    p_mean_data.append(proteomics_mean)
    p_std_data.append(proteomics_std)
    count += 1
    print(f"finish {count} entries")
aggregation = {"improve_sample_id": complete_samples, "transcriptomics_mean": t_mean_data, "transcriptomics_std": t_std_data, "proteomics_mean": p_mean_data, "proteomics_std": p_std_data}
aggregated_df = pd.DataFrame(aggregation)
aggregated_df.to_csv("../data/Cancer/aggregation/stp.csv", index=False)
aggregated_df
```
